# Задача 1. Геометрия детектора

Сфера $R = 2$ м, $N = 40$ ФЭУ с плоскими фотокатодами $r = 10$ см, $a = 0.5$.

**Расстановка ФЭУ.** Используем спираль Фибоначчи: $\cos\theta_i = 1 - 2(i+0.5)/N$ (равные по площади пояса),
$\phi_i = (i+0.5)\cdot\pi(3-\sqrt5)$ («золотой угол»). Координаты хранятся в `numpy`-массиве формы `(N, 3)` —
строка = ФЭУ; это позволяет векторно считать скалярные произведения со всеми ФЭУ сразу.

**Принадлежность точки ФЭУ.** Катод — плоский диск радиуса $r$ с осью $\vec n_k$, вписанный в сферу (край диска лежит на поверхности).
Точка $\vec p$ на сфере принадлежит ФЭУ $k$, если она лежит в сегменте за диском: $|\vec p_T|^2 = p^2 - (\vec p\cdot\vec n_k)^2 \le r^2$ и $\vec p\cdot\vec n_k>0$.
Этим заняты методы `pmt_index` (номер ФЭУ или $-1$) и `is_on_pmt` (да/нет).

```python
import sys
sys.path.insert(0, '../src')   # код проекта
import numpy as np
import matplotlib.pyplot as plt
from detector import Detector

det = Detector(R=2.0, N=40, r=0.10, a=0.5)
print(det.pmt_positions.shape)
det.pmt_positions[:5]
```

## Проверка равномерности

```python
print(f"Мин. расстояние между ФЭУ: {det.min_pmt_distance():.3f} м (2r = {2*det.r} м)")
print(f"Центр масс ФЭУ: {det.pmt_positions.mean(axis=0).round(4)}")
print(f"Доля покрытия поверхности: {det.coverage():.4%}")

d = np.linalg.norm(det.pmt_positions[:, None] - det.pmt_positions[None], axis=2)
np.fill_diagonal(d, np.inf)
plt.figure(figsize=(6, 2.5))
plt.hist(d.min(axis=1), bins=15)
plt.xlabel('расстояние до ближайшего ФЭУ, м'); plt.ylabel('N ФЭУ')
plt.show()
```

## Проверка метода `pmt_index` (Монте-Карло)
Случайные изотропные точки на сфере: доля попавших должна совпасть с аналитическим покрытием.

```python
rng = np.random.default_rng(1)
Mpts = 2_000_000
phi = rng.uniform(0, 2*np.pi, Mpts); ct = rng.uniform(-1, 1, Mpts); st = np.sqrt(1-ct**2)
points = det.R * np.stack([st*np.cos(phi), st*np.sin(phi), ct], axis=1)
idx = det.pmt_index(points)
frac = (idx >= 0).mean()
print(f"MC: {frac:.5f} ± {np.sqrt(frac/Mpts):.5f},  аналитика: {det.coverage():.5f}")
print("is_on_pmt: центр ФЭУ №0 —", det.is_on_pmt(det.pmt_positions[0])[0],
      ", противоположная точка —", det.is_on_pmt(-det.pmt_positions[0])[0])
print("центр каждого ФЭУ принадлежит ему:", np.array_equal(det.pmt_index(det.pmt_positions), np.arange(det.N)))
```

```python
hit = idx >= 0
th, ph = np.arccos(points[:, 2]/det.R), np.arctan2(points[:, 1], points[:, 0])
plt.figure(figsize=(8, 4))
plt.scatter(ph[hit][:20000], np.cos(th[hit][:20000]), s=0.5)
plt.xlabel(r'$\phi$'); plt.ylabel(r'$\cos\theta$'); plt.title('Точки, попавшие на катоды')
plt.show()
```

## Визуализация геометрии

Катоды нарисованы в истинном масштабе.

```python
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

def disk(n, R, r, m=40):
    # контур вписанного катода с осью n
    u = np.cross(n, [0, 0, 1.0]) if abs(n[2]) < 0.9 else np.cross(n, [1.0, 0, 0])
    u /= np.linalg.norm(u)
    v = np.cross(n, u)
    a = np.linspace(0, 2*np.pi, m)[:, None]
    return np.sqrt(R**2 - r**2) * n + r * (np.cos(a) * u + np.sin(a) * v)

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
u, v = np.mgrid[0:2*np.pi:37j, 0:np.pi:19j]
ax.plot_wireframe(det.R*np.cos(u)*np.sin(v), det.R*np.sin(u)*np.sin(v), det.R*np.cos(v),
                  color='0.8', lw=0.4)
ax.add_collection3d(Poly3DCollection([disk(n, det.R, det.r) for n in det.pmt_directions],
                                     facecolor='C1', edgecolor='k', lw=0.5))
ax.set_box_aspect((1, 1, 1))
ax.set_xlabel('x, м'); ax.set_ylabel('y, м'); ax.set_zlabel('z, м')
ax.set_title(f'{det.N} катодов на сфере R = {det.R:g} м')
plt.show()
```
